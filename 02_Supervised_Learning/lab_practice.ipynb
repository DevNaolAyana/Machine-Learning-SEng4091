```python
import pandas as pd
```

```python
data=pd.DataFrame({
    'Gender':['Male','Female','Male','Female'],
    'Admission_Type':['Emergency','Elective','Urgent','Emergency']
})
```

```python
encoded_data=pd.get_dummies(data,columns=['Gender','Admission_Type'],
                            dtype=int)
print(encoded_data)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
data=pd.DataFrame({
    'Medical_Condition':['Diabetes','Cancer','Asthma','Diabetes']
})
```

```python
encoder=LabelEncoder()
data['Condition_Encoded']=encoder.fit_transform(data['Medical_Condition'])
print(data)
```
